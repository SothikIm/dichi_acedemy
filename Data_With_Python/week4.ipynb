```python
import random
print("\t\tGuessing number game")
print("\t\nGuess the number between 1 to 20")
secret_number = random.randint(1, 20)
while(True):
    num = int(input("Enter your number to guess: "))
    if(num == secret_number):
        print(f"Congrat!!! The correct answer is : {secret_number}")
        break
    elif(num < secret_number):
        print("Your number is smaller than the secret number")
    else:
        print("Your number is greater than the secret number")
```
