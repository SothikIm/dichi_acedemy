```python
import numpy as np
import pandas as pd

# Practice 1
df = pd.read_csv("p1_weekly_steps.csv")
print(df)
steps = df["Steps"].to_numpy()
days = df["Day"].to_numpy()

print(f"Total step is {steps.sum()}")
print(f"Day that active the most is {days[steps.argmax()]}")
print(f"Day that active the least is {days[steps.argmin()]}")
print(f"There are {(steps > 10000).sum()} days that above 10000 steps")
```

```python
# Practice 2
df = pd.read_csv("p2_weekly_sales.csv")
print(df)
product = df["Product"].to_numpy()
sales = df[["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]].to_numpy()
day_of_the_week = df.columns[1:]

print("Data for this week: ")
for item, sale in zip(product, sales):
    print(f"\t{item} sold {sale.sum()}")

sale_per_week = sales.sum(axis=1)
print(f"\tThe product that sold the most is: {product[sale_per_week.argmax()]}")
print(f"\tThe product that sold the least is: {product[sale_per_week.argmin()]}")

sale_for_each_day = sales.sum(axis=0)
print(f"\tThe day that sold the most is {day_of_the_week[sale_for_each_day.argmax()]}")

threshold = 200
days = day_of_the_week[sale_for_each_day > threshold]
print(f"\tThe day that above {threshold}: {", ".join(days) if len(days) > 0 else "None"} ")
```

```python
# Practice 3
df = pd.read_csv("p3_monthly_temperatures.csv")
print(df)
cities = df["City"].to_numpy()
temperatures = df.iloc[:, 1:].to_numpy()

for city, temperature in zip(cities, temperatures):
    print(f"{city} have average temperature: {np.round(temperature.mean(), 2)}")
print(f"The warmest city is {cities[temperatures.mean(axis=1).argmax()]}")
print(f"The warmest city is {cities[temperatures.mean(axis=1).argmin()]}")

highest_temp = temperatures.max(axis=1)
lowest_temp = temperatures.min(axis=1)
print(f"The city that have the greatest different temperature is {cities[(highest_temp - lowest_temp).argmax()]}")

difference_AB = np.abs(temperatures[0] - temperatures[1])
difference_AC = np.abs(temperatures[0] - temperatures[2])
difference_BC = np.abs(temperatures[1] - temperatures[2])

print(f"The difference bettween city A and B is {difference_AB.mean()}")
print(f"The difference bettween city A and C is {difference_AC.mean()}")
print(f"The difference bettween city C and B is {difference_BC.mean()}")

threshold = 200
matching_city = cities[np.any(temperatures > 25, axis=1)]
print(f"The city the experence above 25 {", ".join(matching_city) if len(matching_city) else "None"}")
```
