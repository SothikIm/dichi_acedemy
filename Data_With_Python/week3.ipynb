```python
# Simple Calculator 
num1 = int(input("Enter the frist number: "))
num2 = int(input("Enter the second number: "))
operator = input("Enter an operation (+ - * /): ")
if operator == '+':
    print(f"The result is : {num1 + num2}")
elif operator == '-':
    print(f"The result is : {num1 - num2}")
elif operator == '*':
    print(f"The result is : {num1 * num2}")
elif operator == '/':
    if num2 == 0:
        print("Undifine behavior, the second number must be greater than 0")
    else:
        print(f"The result is : {num1 / num2}")
else:
    print("Error: Invalid operation enter")
```

```python
# Discount calculator
total_amount = float(input("Enter the total amount of purchase: $"))
if total_amount > 100:
    total_amount = (total_amount * 90) / 100
elif total_amount <= 100:
    total_amount = (total_amount * 95) / 100
print(f"The final amount after discount: {total_amount}")
```
