```python
def triangle_area(width, height):
    return (width * height) / 2

print(triangle_area(10, 5))
print(triangle_area(7, 3))
```

```python
def celsius_to_fahrenheit(celsius):
    print(f"0°C is equal to {(celsius * 9 / 5) + 32}°F")

celsius_to_fahrenheit(0)
celsius_to_fahrenheit(100)
celsius_to_fahrenheit(37)
```

```python
def find_large(num1, num2, num3):
    largest = max(num1, num2, num3)
    print(f"The largest number among {num1} {num2} {num3} is {largest}")
find_large(5, 10, 3)
find_large(-4, -1, -7)
find_large(8, 8, 2)
```
